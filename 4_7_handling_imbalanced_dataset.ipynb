```python
import numpy as np
import pandas as pd
```

```python
data=pd.read_csv('/content/Credit.csv')
data.head()
```

```python
data.tail()
```

```python
#distribution of the two classes
data['Student'].value_counts()
```

```python
#separating student and non student
student=data[data['Student']=='Yes']
non_student=data[data['Student']=='No']
print(student.shape)
print(non_student.shape)
```

```python
#under sample create equal distribution
student_sample=non_student.sample(n=40)
print(student_sample.shape)
```

```python
#concate the two dataframe
new_data=pd.concat([student_sample,student] ,axis=0)
new_data.head()
```

```python
new_data['Student'].value_counts()
```
