vector dot product
a=([1,2])
b=([2,3])

```python
import numpy as np
a=np.array([1,2])
b=np.array([2,1])

a_dot_b=np.dot(a,b)
print(a_dot_b)
```

cross product

```python
a_cross_b=np.cross(a,b)
print(a_cross_b)
```

projection of  A vector on v

```python
a=np.array([2,3])
v=np.array([4,5])
magnitude=np.sqrt(sum(v**2))
projection=np.dot(a,v)/magnitude**2*v
print(projection)
```
