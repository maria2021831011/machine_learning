```python
new_data=pd.read_csv('/content/train.csv')
new_data.shape
```

```python
# replacing the null values with empty string
new_data = new_data.fillna('')
```

```python
# merging the author name and news title
new_data['content'] = new_data['Name']+' '+new_data['Sex']
```

```python
print(new_data['content'])
```

```python
# separating the data & survived
x=new_data.drop(columns='Survived' ,axis=1)
y=new_data['Survived']
print(x)
```

```python
print(y)
```

```python
#Stemming:

#Stemming is the process of reducing a word to its Root word

#example: actor, actress, acting --> act
from nltk.stem.porter import PorterStemmer
port_stem=PorterStemmer()
```

```python
def stemming(content):
    stemmed_content = re.sub('[^a-zA-Z]',' ',content)
    stemmed_content = stemmed_content.lower()
    stemmed_content = stemmed_content.split()
    stemmed_content = [port_stem.stem(word) for word in stemmed_content
                       if not word in stopwords.words('english')]
    stemmed_content = ' '.join(stemmed_content)
    return stemmed_content
```

```python
new_data['content'] = new_data['content'].apply(stemming)
```

```python
print(new_data['content'])
```

```python
#separating the data and label
x= new_data['content'].values
y= new_data['Survived'].values
```

```python
# convert the textual data to Feature Vectors
vectorizer = TfidfVectorizer()
```

```python
vectorizer.fit(x)
x=vectorizer.transform(x)
print(x)
```
