```python
#dependency
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

data collection and data procecing

```python
#pandas read dataframe
sonar_data=pd.read_csv('/content/Copy of sonar data.csv',header=None)
```

```python
sonar_data.head()
```

```python
sonar_data.shape
```

```python
sonar_data[60].value_counts() # statisticaal measure
```

```python
sonar_data.groupby(60).mean()
```

```python
#spliting
X=sonar_data.drop(columns=60,axis=1)
Y=sonar_data[60]
print(X)
print(Y)
```

```python
# train data and test data
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.1,stratify=Y,random_state=1)
```

```python
print(X.shape,X_train.shape,X_test.shape)
```

```python
print(X_train)
print(Y_train)
```

model training logistic regression

```python
model=LogisticRegression()
model.fit(X_train,Y_train)
```

EVALUTION

```python
x_accuracy=model.predict(X_train)
training_data_accuracy=accuracy_score(x_accuracy,Y_train)
print(training_data_accuracy)
```

```python
x_accuracy=model.predict(X_test)
testing_data_accuracy=accuracy_score(x_accuracy,Y_test)
print(testing_data_accuracy)
```
